```python
import pandas as pd

df = pd.read_csv("../data/processed/cleaned_matches.csv")
df["date"] = pd.to_datetime(df["date"])

df = df.sort_values("date")
df.head()
```

```python
# Store historical team statistics
from collections import defaultdict
team_stats = defaultdict(
    lambda: {
        "matches": 0,
        "wins": 0,
        "draws": 0,
        "losses": 0,
        "goals_scored": 0,
        "goals_conceded": 0,
    }
)

# Feature storage
home_matches_played = []
away_matches_played = []

home_win_rate = []
away_win_rate = []

home_draw_rate = []
away_draw_rate = []

home_loss_rate = []
away_loss_rate = []

home_avg_goals_scored = []
away_avg_goals_scored = []

home_avg_goals_conceded = []
away_avg_goals_conceded = []
```

```python
for _, row in df.iterrows():

    home = row["home_team"]
    away = row["away_team"]

    home_stats = team_stats[home]
    away_stats = team_stats[away]


    # HOME TEAM FEATURES
    hm = home_stats["matches"]

    home_matches_played.append(hm)

    if hm > 0:
        home_win_rate.append(home_stats["wins"] / hm)
        home_draw_rate.append(home_stats["draws"] / hm)
        home_loss_rate.append(home_stats["losses"] / hm)

        home_avg_goals_scored.append(
            home_stats["goals_scored"] / hm
        )

        home_avg_goals_conceded.append(
            home_stats["goals_conceded"] / hm
        )

    else:
        home_win_rate.append(0)
        home_draw_rate.append(0)
        home_loss_rate.append(0)

        home_avg_goals_scored.append(0)
        home_avg_goals_conceded.append(0)

    # AWAY TEAM FEATURES
    am = away_stats["matches"]

    away_matches_played.append(am)

    if am > 0:
        away_win_rate.append(away_stats["wins"] / am)
        away_draw_rate.append(away_stats["draws"] / am)
        away_loss_rate.append(away_stats["losses"] / am)

        away_avg_goals_scored.append(
            away_stats["goals_scored"] / am
        )

        away_avg_goals_conceded.append(
            away_stats["goals_conceded"] / am
        )

    else:
        away_win_rate.append(0)
        away_draw_rate.append(0)
        away_loss_rate.append(0)

        away_avg_goals_scored.append(0)
        away_avg_goals_conceded.append(0)

    # UPDATE TEAM STATS AFTER FEATURE CREATION
    home_score = row["home_score"]
    away_score = row["away_score"]

    # Matches played

    team_stats[home]["matches"] += 1
    team_stats[away]["matches"] += 1

    # Goals

    team_stats[home]["goals_scored"] += home_score
    team_stats[home]["goals_conceded"] += away_score

    team_stats[away]["goals_scored"] += away_score
    team_stats[away]["goals_conceded"] += home_score

    # Results

    if home_score > away_score:

        team_stats[home]["wins"] += 1
        team_stats[away]["losses"] += 1

    elif home_score < away_score:

        team_stats[away]["wins"] += 1
        team_stats[home]["losses"] += 1

    else:

        team_stats[home]["draws"] += 1
        team_stats[away]["draws"] += 1
```

```python
df["home_matches_played"] = home_matches_played
df["away_matches_played"] = away_matches_played

df["home_win_rate"] = home_win_rate
df["away_win_rate"] = away_win_rate

df["home_draw_rate"] = home_draw_rate
df["away_draw_rate"] = away_draw_rate

df["home_loss_rate"] = home_loss_rate
df["away_loss_rate"] = away_loss_rate

df["home_avg_goals_scored"] = home_avg_goals_scored
df["away_avg_goals_scored"] = away_avg_goals_scored

df["home_avg_goals_conceded"] = home_avg_goals_conceded
df["away_avg_goals_conceded"] = away_avg_goals_conceded
#First matches for all team has 0 avg
```

```python
df["win_rate_diff"] = (
    df["home_win_rate"]
    - df["away_win_rate"]
)

df["draw_rate_diff"] = (
    df["home_draw_rate"]
    - df["away_draw_rate"]
)

df["loss_rate_diff"] = (
    df["home_loss_rate"]
    - df["away_loss_rate"]
)

df["goals_scored_diff"] = (
    df["home_avg_goals_scored"]
    - df["away_avg_goals_scored"]
)

df["goals_conceded_diff"] = (
    df["home_avg_goals_conceded"]
    - df["away_avg_goals_conceded"]
)

df["matches_played_diff"] = (
    df["home_matches_played"]
    - df["away_matches_played"]
)
df
```

```python
feature_cols = [
    "home_win_rate",
    "away_win_rate",
    "home_avg_goals_scored",
    "away_avg_goals_scored",
    "win_rate_diff",
]

df[feature_cols].tail(20)
```

```python
df.to_csv("../data/processed/features_v1.csv",
            index=True)
```

# Recent Form

```python
from collections import defaultdict, deque

recent_results = defaultdict(lambda: deque(maxlen=5))
recent_goals_scored = defaultdict(lambda: deque(maxlen=5))
recent_goals_conceded = defaultdict(lambda: deque(maxlen=5))

# Feature lists

home_last5_win_rate = []
away_last5_win_rate = []

home_last5_avg_goals_scored = []
away_last5_avg_goals_scored = []

home_last5_avg_goals_conceded = []
away_last5_avg_goals_conceded = []
```

```python
for _, row in df.iterrows():

    home = row["home_team"]
    away = row["away_team"]

    # HOME TEAM FEATURES
    if len(recent_results[home]) > 0:

        home_last5_win_rate.append(
            sum(recent_results[home]) / len(recent_results[home])
        )

        home_last5_avg_goals_scored.append(
            sum(recent_goals_scored[home])
            / len(recent_goals_scored[home])
        )

        home_last5_avg_goals_conceded.append(
            sum(recent_goals_conceded[home])
            / len(recent_goals_conceded[home])
        )

    else:

        home_last5_win_rate.append(0)
        home_last5_avg_goals_scored.append(0)
        home_last5_avg_goals_conceded.append(0)

    # AWAY TEAM FEATURES
    if len(recent_results[away]) > 0:

        away_last5_win_rate.append(
            sum(recent_results[away]) / len(recent_results[away])
        )

        away_last5_avg_goals_scored.append(
            sum(recent_goals_scored[away])
            / len(recent_goals_scored[away])
        )

        away_last5_avg_goals_conceded.append(
            sum(recent_goals_conceded[away])
            / len(recent_goals_conceded[away])
        )

    else:

        away_last5_win_rate.append(0)
        away_last5_avg_goals_scored.append(0)
        away_last5_avg_goals_conceded.append(0)
    
    # UPDATE HISTORY AFTER FEATURE CREATION
    home_score = row["home_score"]
    away_score = row["away_score"]

    # Match result encoding
    # Win = 1
    # Draw = 0.5
    # Loss = 0

    if home_score > away_score:

        recent_results[home].append(1)
        recent_results[away].append(0)

    elif home_score < away_score:

        recent_results[home].append(0)
        recent_results[away].append(1)

    else:

        recent_results[home].append(0.5)
        recent_results[away].append(0.5)

    # Goals

    recent_goals_scored[home].append(home_score)
    recent_goals_conceded[home].append(away_score)

    recent_goals_scored[away].append(away_score)
    recent_goals_conceded[away].append(home_score)
```

```python
df["home_last5_win_rate"] = home_last5_win_rate
df["away_last5_win_rate"] = away_last5_win_rate

df["home_last5_avg_goals_scored"] = (
    home_last5_avg_goals_scored
)

df["away_last5_avg_goals_scored"] = (
    away_last5_avg_goals_scored
)

df["home_last5_avg_goals_conceded"] = (
    home_last5_avg_goals_conceded
)

df["away_last5_avg_goals_conceded"] = (
    away_last5_avg_goals_conceded
)
```

```python
df["last5_win_rate_diff"] = (
    df["home_last5_win_rate"]
    - df["away_last5_win_rate"]
)

df["last5_goals_scored_diff"] = (
    df["home_last5_avg_goals_scored"]
    - df["away_last5_avg_goals_scored"]
)

df["last5_goals_conceded_diff"] = (
    df["home_last5_avg_goals_conceded"]
    - df["away_last5_avg_goals_conceded"]
)
```

```python
recent_cols = [
    "home_last5_win_rate",
    "away_last5_win_rate",
    "last5_win_rate_diff"
]

df[recent_cols].tail(20)
```
