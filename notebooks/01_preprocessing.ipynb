```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"..\data\raw\results.csv")
```

```python
df.isnull().sum()
#Existance of null values in scores
#Matches that are yet to happen
df = df.dropna(subset=["home_score","away_score"])
df
```

```python
#Standardize dates
df["date"] = pd.to_datetime(df["date"])
df["year"] = df["date"].dt.year 
df["month"] = df["date"].dt.month 
df
```

```python
former = pd.read_csv(r"..\data\raw\former_names.csv")
```

```python
#Replace older country names with newer ones
name_map = dict(zip(former["former"],former["current"]))

df["home_team"] = df["home_team"].replace(name_map)
df["away_team"] = df["away_team"].replace(name_map)
df
```

```python
old_matches = df[df["date"].dt.year>=2000]
df = df[df["date"].dt.year>=2000]
df = df.reset_index(drop=True)
df
```

```python
def result(row):
    if row["home_score"] > row["away_score"]:
        return "Home Win"
    elif row["home_score"] < row["away_score"]:
        return "Away Win"
    else:
        return "Draw"

df["result"] = df.apply(result,axis=1)
df.head()
```

```python
#Prevent data leakage
df.drop(["home_score","away_score"],axis=1)
```

```python
df.to_csv(
    "../data/processed/cleaned_matches.csv",
    index=False
)
```
